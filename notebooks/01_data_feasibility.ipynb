## Feasibility verdict

**Decision: GO**

The Toronto pilot is feasible with the current open data.

- The general Traffic Collisions dataset contains 825,265 unique collision events and is suitable for overall collision context, but it does not distinguish serious injuries from other injuries.
- The actively maintained City of Toronto KSI dataset provides the required KSI definition and contains 20,771 person-level records representing 7,620 unique collision events.
- Event-level road context is internally consistent for `traffictl` and `accloc`.
- 7,619 of 7,620 KSI collision events have complete latitude and longitude.
- Using the strict pilot definition (`Traffic Signal` + explicit intersection location categories), 2,252 unique KSI collision events are available.

**Constraint:** the KSI dataset, not the general Traffic Collisions dataset, should be used as the source of the KSI outcome. `Intersection-Related` is excluded from the strict pilot definition until its interpretation is justified more clearly.
